```python
import pandas as pd

# Exemple de DataFrame avec des chaînes de caractères
df = pd.DataFrame({
    'col1': ['a', 'b', 'c'],
    'col2': ['x', 'y', 'z'],
    'col3': ['1', '2', '3']
})

# Création d'une nouvelle colonne en joignant les valeurs ligne par ligne avec '_'
df['joined'] = df[['col1','col2']].astype(str).apply(lambda x: '_'.join(x), axis=1)

display(df)
```
