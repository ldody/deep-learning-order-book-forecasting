```python
import os, sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {'split':['Train_Test_Split','Optuna_Split','Train_Evaluation_Test_Split',''], 
        'Train':[80, 80, 50,0], 
        'Evaluation':[0,0,30,0],
        'Test':[20,20,20,0]}


df = pd.DataFrame(data).set_index('split').iloc[::-1]
df.index.name = None
df.plot.barh(stacked=True, figsize=(16,8))
plt.annotate('SHUFFLED', xy=(40, 0.6), xytext=(40, 0.4), xycoords='data', 
            fontsize=10*1.5, ha='center', va='bottom',
            arrowprops=dict(arrowstyle='-[, widthB=23.0, lengthB=1', lw=2.0, color='k'))

plt.annotate('SHUFFLED', xy=(40, 1.6), xytext=(40, 1.4), xycoords='data', 
            fontsize=10*1.5, ha='center', va='bottom',
            arrowprops=dict(arrowstyle='-[, widthB=23.0, lengthB=1', lw=2.0, color='k'))

plt.legend(ncols=3, loc='lower center', bbox_to_anchor=(0.5, 0.1), fontsize=15)

for idx, row in df[1:].iterrows():
    p = 0
    for i in row:
        if i != 0:
            plt.text(s=f'{i} %', x=p+i/2, y=df.index.get_loc(idx), va='center', ha='center', fontsize=15)
        
        p += i
    
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_visible(False)

fig = plt.gcf()

ax.get_xaxis().set_visible(False)
plt.xlim(0, 100)
plt.yticks(fontsize=15)

plt.title('Data Split', fontsize=20)

ax_right = ax.twinx()
ax_right.set_ylim(ax.get_ylim())
ax_right.set_yticks(ax.get_yticks())
ax_right.set_yticklabels([label.get_text() for label in ax.get_yticklabels()], fontsize=15)
ax_right.tick_params(axis='y', labelleft=False, labelright=True, colors='white')

for spine in ax_right.spines.values():
    spine.set_visible(False)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Données pour l'exemple
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Création de la figure et des axes avec une taille personnalisée
fig, ax = plt.subplots(figsize=(8, 6))  # Modifier la taille de la figure

# Tracer les données
ax.plot(x, y)

# Ajuster les marges et utiliser tight_layout
plt.subplots_adjust(left=0.15, right=0.85, top=0.9, bottom=0.1)
plt.tight_layout()

# Afficher le graphique
plt.show()
```

```python
data = {'split':['Train_Test_Split','Optuna_Split','Train_Evaluation_Test_Split',''], 
        'Train':[80, 80, 50,0], 
        'Evaluation':[0,0,30,0],
        'Test':[20,20,20,0]}


df = pd.DataFrame(data).set_index('split').iloc[::-1]
df.index.name = None
df.plot.barh(
    stacked=True, figsize=(16,8),
    color=["0.85", "0.65", "0.35"],   # gris clair -> foncé
    edgecolor="black", linewidth=1.0
)
plt.annotate('SHUFFLED', xy=(40, 0.6), xytext=(40, 0.4), xycoords='data', 
            fontsize=10*1.5, ha='center', va='bottom',
            arrowprops=dict(arrowstyle='-[, widthB=23.0, lengthB=1', lw=2.0, color='k'))

plt.annotate('SHUFFLED', xy=(40, 1.6), xytext=(40, 1.4), xycoords='data', 
            fontsize=10*1.5, ha='center', va='bottom',
            arrowprops=dict(arrowstyle='-[, widthB=23.0, lengthB=1', lw=2.0, color='k'))

plt.legend(ncols=3, loc='lower center', bbox_to_anchor=(0.5, 0.1), fontsize=15)

for idx, row in df[1:].iterrows():
    p = 0
    for i in row:
        if i != 0:
            plt.text(
    s=f'{i} %',
    x=p+i/2,
    y=df.index.get_loc(idx),
    va='center',
    ha='center',
    fontsize=15,
    bbox=dict(facecolor='white', edgecolor='none', pad=5)
)
        
        p += i
    
ax = plt.gca()
for p in ax.patches:
    if p.get_width() == 0:
        p.set_linewidth(0)
        
for spine in ax.spines.values():
    spine.set_visible(False)

fig = plt.gcf()

ax.get_xaxis().set_visible(False)
plt.xlim(0, 100)
plt.yticks(fontsize=15)

plt.title('Data Split', fontsize=20)

ax_right = ax.twinx()
ax_right.set_ylim(ax.get_ylim())
ax_right.set_yticks(ax.get_yticks())
ax_right.set_yticklabels([label.get_text() for label in ax.get_yticklabels()], fontsize=15)
ax_right.tick_params(axis='y', labelleft=False, labelright=True, colors='white')

for spine in ax_right.spines.values():
    spine.set_visible(False)
```
