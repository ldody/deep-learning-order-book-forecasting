```python
import os, sys
import pandas as pd

cwd = os.getcwd()
root_path = cwd
while '.venv' not in os.listdir(root_path):
    root_path =  os.path.dirname(root_path)
data_path = os.path.join(root_path,'data')
display(os.listdir(data_path))
```

```python
df = pd.read_csv(os.path.join(data_path,'assets_DB.csv'), index_col=0)
display(df)
```

```python
df = df.drop(['optimization','process'], axis=1)
df['file'] = df[['ISIN','data','model']].apply(lambda x: '_'.join(x), axis=1)
df['GSA'] = 'PENDING'
display(df)
```

```python
df.to_csv('assets_GSA.csv')
```
