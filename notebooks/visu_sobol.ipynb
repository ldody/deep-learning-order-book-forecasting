```python
import os, sys
import pandas as pd
import numpy as np

src_path = os.getcwd()
while '.venv' not in os.listdir(src_path):
    src_path =  os.path.dirname(src_path)
GSA_path = os.path.join(src_path,'results','GSA')
display(src_path)

cwd = os.getcwd()

ls_files_s = [f for f in os.listdir(GSA_path) if 'LSA' not in f]
n_ohlcv = len([f for f in ls_files_s if 'OHLCV' in f])
display(ls_files_s, n_ohlcv)
```

```python
df = pd.DataFrame()
m = ['CNN','LSTM']

t_df = pd.DataFrame()
top10 = pd.DataFrame()

for f in ls_files_s:
    tmp = pd.read_parquet(os.path.join(GSA_path, f))
    tmp['model'] = f.split('.')[0]
    tmp['model'] = tmp['model'].apply(lambda x: '_'.join([i for i in x.split('_') if i in m]))
    df = pd.concat([df,tmp])
    
#display(df)
for m in ['CNN','LSTM','CNN_LSTM']:
    for s in ['S1','ST']:
        func1(df[df['model'] == m], s)
```

```python
def func1(df, val):
    df_res_ST = df.reset_index().groupby(['index','target'], as_index=False).sum()

    ohlcv = ['Open','High','Low','Volume']

    df_res_ST.loc[~df_res_ST['index'].isin(ohlcv),['ST','ST_conf','S1','S1_conf']] = df_res_ST.loc[~df_res_ST['index'].isin(ohlcv),['ST','ST_conf','S1','S1_conf']]/len(ls_files_s)
    df_res_ST.loc[df_res_ST['index'].isin(ohlcv),['ST','ST_conf','S1','S1_conf']] = df_res_ST.loc[df_res_ST['index'].isin(ohlcv),['ST','ST_conf','S1','S1_conf']]/n_ohlcv

    df_res_ST['row'] = df_res_ST['index'].apply(lambda x: x.split('_')[-1])
    df_res_ST['row'] = df_res_ST['row'].astype(int)
    df_res_ST['row'] = df_res_ST['row'] - 100
    df_res_ST['index'] = df_res_ST['index'].apply(lambda x: '_'.join(x.split('_')[:-1]))
    df_res_ST = df_res_ST.sort_values('index')

    pivot_table = df_res_ST.pivot(index=['row','target'], columns='index', values=val)
    #display(pivot_table)
    func2(pivot_table)
```

```python
def func2(pivot_table):
    pivot_table = pivot_table.fillna(0)
    t = pivot_table.reset_index().groupby('row').mean()
    t = t.drop(['target'], axis=1)
    #display(pivot_table.reset_index().groupby('row').mean())
    #display(t.sort_index())
    
    t = t.sort_index(axis=1)
    
    def sort_key(col):
        parts = col.split('_')
        if len(parts) == 1:
            return (col, -1)
        try:
            prefix, number = parts[:-1], parts[-1]
            return ('_'.join(prefix), int(number))
        except:
            return (col, float('inf'))  # place en fin ou adapte selon tes besoins

    sorted_columns = sorted(t.columns, key=sort_key)
    t = t[sorted_columns]
    
    func3(t)
    func4(t)
    func5(t)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

def func3(t):
    plt.figure(figsize=(20, 10))
    #sns.heatmap(pivot_table, cmap="Reds")
    sns.heatmap(t, cmap="Reds")
```

```python
from scipy import stats

def func4(t):
    global t_df
    
    t_sum = t.dropna().mean(axis=1)
    
    t_df = pd.concat([t_df, pd.DataFrame(t_sum)], ignore_index=True, axis=1)
    
    cs = t_sum.cumsum()

    x = t_sum.index
    y = t_sum
    
    cs_norm = cs / cs.max() * y.max()
    y = cs_norm
    ln_y = np.log(y)

    # Régression linéaire sur x et ln(y)
    slope, intercept, r_value, p_value, std_err = stats.linregress(x, ln_y)

    # Récupérer les paramètres a et b
    b = slope
    a = np.exp(intercept)

    #print(f"a = {a:.3f}, b = {b:.3f}")

    # Calcul des valeurs ajustées pour affichage
    y_fit = a * np.exp(b * x)

    

    plt.figure(figsize=(16, 8))
    plt.plot(t_sum)
    plt.plot(x, y_fit)
    plt.plot(cs_norm)

    plt.show()
```

```python
def func5(t):
    global top10
    global m
    global s
    
    tmp = t.mean(axis=0).sort_values(ascending=False).to_frame().reset_index()[:10]
    name_m = '+'.join(m.split('_'))
    tmp.columns = [f'{name_m} {s}', f'{name_m} {s}']
    display(tmp)
    top10 = pd.concat([top10, tmp], ignore_index=False, axis=1)
    
    display(t.mean(axis=0).sort_values().to_frame().T)
    display(t.mean(axis=0).sort_values().to_frame().T['Close'])
    check = t.mean(axis=0).sort_values().to_frame().T
    plt.figure(figsize=(16, 8))
    sns.heatmap(t.mean(axis=0).sort_values().to_frame().T, cmap="Reds")
```
