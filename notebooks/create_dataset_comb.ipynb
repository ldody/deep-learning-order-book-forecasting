```python
import os, sys
import pandas as pd

cwd = os.getcwd()

path_data = os.path.join(os.path.dirname(cwd), 'data')
display(path_data)
```

```python
display(os.listdir(path_data))
```

```python
df = pd.read_csv(os.path.join(path_data, 'assets.csv'), index_col=0, usecols=[0,1,2])
ls_data = ['OHLCV','FOB','OHLCV_FOB']
ls_model = ['CNN','LSTM','CNN_LSTM']
df['data'] = [ls_data] * len(df)
df['model'] = [ls_model] * len(df)
df['optimization'] = 0
df['process'] = False

df = df.explode(['data']).explode('model').reset_index(drop=True)
display(df)
```

```python
df.to_csv(os.path.join(path_data, 'assets_DB.csv'))
```
